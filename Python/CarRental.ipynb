```python
# Import Libraries
import pandas as pd
import numpy as np
import random
from datetime import datetime, timedelta

print("Libraries imported — Ready")

#STEP 1- Set Seed
np.random.seed(178)  # numpy seed
random.seed(178)     # random seed

print("Seed set 178 — Every time same random — So file same")

# STEP 2-Define How Many — How big dataset
n_trips = 100000   # 100K trips
n_cars = 650       # 650 cars
n_customers = 8500 # 8500 customers
print(f"We will create {n_trips} trips, {n_cars} cars, {n_customers} customers")

#STEP 3- Make Lists
car_models = ['【entity-Hyundai¦canonical_name=Hyundai】 Accent', '【entity-Toyota¦canonical_name=Toyota】 Camry', '【entity-Nissan¦canonical_name=Nissan】 Sunny', 'Kia Picanto', '【entity-Toyota¦canonical_name=Toyota】 Yaris', '【entity-Hyundai¦canonical_name=Hyundai】 Tucson', 'Kia Cerato']

locations = ['Deira', 'Bur Dubai', 'Marina', 'Downtown', 'Business Bay', 'JLT', 'Al Barsha', 'International City', 'Silicon Oasis', 'Sports City', 'Motor City', 'Discovery Gardens']

channels = ['Google Ads', 'Instagram', 'Referral', 'Organic Search', 'Facebook Ads', 'Direct', 'Partnerships']

payment = ['Card', '【entity-Apple¦canonical_name=Apple】 Pay', 'Wallet', 'Cash']

status_list = ['Completed']*93 + ['Cancelled']*7

print(f"Car models: {car_models}")
print(f"Locations: {locations}")
print(f"Channels: {channels}")
print(f"Status list: 93% Completed, 7% Cancelled")

#STEP 4- Make IDs — Unique names
car_ids = [f"CAR_{i:04d}" for i in range(1, n_cars+1)]

customer_ids = [f"CUST_{i:05d}" for i in range(1, n_customers+1)]

print(f"First 3 car IDs: {car_ids[:3]}")
print(f"Last 3 car IDs: {car_ids[-3:]}")
print(f"First 3 customer IDs: {customer_ids[:3]}")
print(f"Total cars: {len(car_ids)}, Total customers: {len(customer_ids)}")

# STEP 5-Generate Data — Loop 100K times
data = []  # Empty list — Will add 100K rows here
start_date = datetime(2024, 1, 1)  # Start date Jan 1 2024 — 20 months data till Aug 2025

for i in range(1, n_trips+1):  # Loop from 1 to 100000 — 100K times
    # Each time loop makes 1 trip — 16 columns

    trip_id = f"TRIP_{i:06d}"  # TRIP_000001, TRIP_000002 ... 06d = 6 digits

    car_id = random.choice(car_ids)  # Pick random car from 650 — Like =RANDBETWEEN in Excel
    car_model = random.choice(car_models)  # Pick random model from 7 — Accent or Camry etc.

    pickup = random.choice(locations)  # Pick random pickup location — Deira or Marina
    dropoff = random.choice(locations)  # Pick random dropoff — Can be same or different

    customer_id = random.choice(customer_ids)  # Pick random customer from 8500

    channel = random.choice(channels)  # Pick random channel — Google Ads or Instagram etc.

    trip_date = start_date + timedelta(days=random.randint(0, 600))  # Random date — Jan 2024 to Aug 2025 — 600 days

    duration = random.randint(15, 600)  # Trip duration 15 minutes to 600 minutes (10 hours) — Realistic
    distance = round(random.uniform(5, 300), 1)  # Distance 5 km to 300 km — Realistic
    revenue = round(random.uniform(40, 600), 2)  # Revenue AED 40 to 600 per trip — Real Udrive pricing
    fuel = round(revenue * random.uniform(0.08, 0.15), 2)  # Fuel cost 8% to 15% of revenue
    net = round(revenue - fuel, 2)  # Net revenue = revenue - fuel
    rating = random.randint(3, 5)  # Rating 3,4,5 — Usually good
    trip_status = random.choice(status_list)  # 93% Completed, 7% Cancelled — From status_list
    pay = random.choice(payment)  # Payment method

    # Add this trip to data list — Like adding row in Excel
    data.append([trip_id, car_id, car_model, pickup, dropoff, customer_id, channel, trip_date.date(), duration, distance, revenue, fuel, net, rating, trip_status, pay])

print(f"Generated {len(data)} trips — First trip: {data[0]}")
print(f"Last trip: {data[-1]}")

# STEP 7: Make DataFrame — Excel table in Python — 16 columns
columns = ['trip_id','car_id','car_model','pickup_location','dropoff_location','customer_id','acquisition_channel','trip_date','trip_duration_min','distance_km','revenue_aed','fuel_cost_aed','net_revenue_aed','rating','trip_status','payment_method']
# Column names — Like Excel header — 16 columns

df = pd.DataFrame(data, columns=columns)
# Make DataFrame from data list — data has 100K rows — columns has 16 names — Now df is Excel table 100K x 16

print(f"DataFrame created — Shape {df.shape} — {df.shape[0]} rows, {df.shape[1]} columns")
print(df.head(3))  # Show first 3 rows — Like Excel top 3 rows

# STEP 8: Save to CSV — Save Excel table to file — So you can download
df.to_csv('udrive_trips_100k_preetha.csv', index=False)
# Save df to file named udrive_trips_100k_preetha.csv — index=False means don't save row numbers

print(f"✅ File saved — udrive_trips_100k_preetha.csv")
print(f"Total Revenue AED {df['revenue_aed'].sum():,.0f} — Should be ~58M")
print(f"Cars {df['car_id'].nunique()} — Customers {df['customer_id'].nunique()}")
print(f"Status: {df['trip_status'].value_counts().to_dict()}")
print(f"Top Location: {df['pickup_location'].value_counts().head(1).to_dict()}")
print(f"Top Model: {df['car_model'].value_counts().head(1).to_dict()}")

# STEP 9: Check file
import os
size = os.path.getsize('udrive_trips_100k_preetha.csv') / (1024*1024)
print(f"File size {size:.1f} MB — Ready to use in Excel, PostgreSQL, Python, Tableau, Power BI")
```
